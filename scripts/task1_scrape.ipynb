```python
from google_play_scraper import reviews, Sort
import pandas as pd

APP_PACKAGE = "com.combanketh.mobilebanking"
```

```python
all_reviews = []
next_token = None
while len(all_reviews) < 400:
    rv, next_token = reviews(APP_PACKAGE, lang="en", country="us", sort=Sort.NEWEST, count=200, continuation_token=next_token)
    all_reviews += [{"review_text": r["content"], "rating": r["score"], "date": r["at"].strftime("%Y-%m-%d")} for r in rv]
    if not next_token: break
```

```python
df = pd.DataFrame(all_reviews)
df = df.drop_duplicates(subset=["review_text", "rating", "date"]).dropna()
print("Dates:", df["date"].min(), "to", df["date"].max(), "| Total reviews:", len(df))
```

```python
df.to_csv("combank_reviews.csv", index=False, quoting=1)  # quoting=1 → csv.QUOTE_ALL
print("Saved combank_reviews.csv ✔")
```
